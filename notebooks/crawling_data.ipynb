```python
import os
import pandas as pd

# Folder tempat file search_results disimpan
folder_path = "/content/tweets-data/"  # Sesuaikan dengan lokasi folder Anda
output_filename = os.path.join(folder_path, "combined_search_results.csv")  # File output hasil penggabungan

# Inisialisasi daftar untuk menyimpan data
all_data = []

# Iterasi untuk mencari file dengan nama berurutan
index = 0
while True:
    filename = f"search_results{index}.csv"  # Pola nama file
    file_path = os.path.join(folder_path, filename)

    if os.path.exists(file_path):  # Jika file ada, baca dan tambahkan ke daftar
        try:
            df = pd.read_csv(file_path)
            all_data.append(df)
            print(f"File {filename} berhasil dibaca dan ditambahkan.")
        except Exception as e:
            print(f"Gagal membaca file {filename}: {e}")
    else:  # Jika file tidak ada, hentikan iterasi
        break

    index += 1

# Jika ada data, gabungkan menjadi satu DataFrame
if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)
    combined_df.to_csv(output_filename, index=False)
    print(f"Semua file berhasil digabungkan dan disimpan ke '{output_filename}'.")
else:
    print("Tidak ada file yang ditemukan untuk digabungkan.")
```

```python
import pandas as pd
import os

# Nama file input (berisi tweet URLs) dan file output untuk komentar
input_filename = "/content/tweets-data/combined_search_results.csv"  # File input berisi URL postingan
output_filename = 'hasil_komentar.csv'  # File output hasil gabungan komentar
twitter_auth_token = 'f50fc2f32be66abbc3e89c1a018d236f067ea76a'  # Ganti dengan token Anda

# Parameter untuk tweet-harvest
comments_limit = 2000  # Jumlah komentar yang diambil per postingan

# Membaca file input berisi URL postingan
try:
    tweet_urls_df = pd.read_csv(input_filename)
    tweet_urls = tweet_urls_df['tweet_url']  # Pastikan kolom ini ada di file input
except FileNotFoundError:
    print(f"File '{input_filename}' tidak ditemukan.")
    exit()

# Cek apakah file output sudah ada, jika ada hapus dulu untuk penggabungan ulang
if os.path.exists(output_filename):
    os.remove(output_filename)

# Iterasi mengambil komentar dari setiap URL
for idx, url in enumerate(tweet_urls):
    print(f"Mengambil komentar dari postingan {idx + 1}/{len(tweet_urls)}: {url}")
    temp_filename = f"temp_comments_{idx}.csv"  # File sementara untuk setiap URL

    # Jalankan tweet-harvest untuk mengambil komentar
    os.system(f'npx --yes tweet-harvest@latest -o {temp_filename} -l {comments_limit} --token {twitter_auth_token} --thread {url}')

```

```python
# Optional (menambah Hasil Manual Crawling)

import os
import shutil

# Hubungkan Google Drive ke Colab
from google.colab import drive
drive.mount('/content/drive')

# Folder sumber dan tujuan
drive_folder = "/content/drive/MyDrive/Colab Notebooks/manualCrawling/"
destination_folder = "/content/tweets-data"

# Pastikan folder tujuan ada, jika tidak, buat folder
if not os.path.exists(destination_folder):
    os.makedirs(destination_folder)

# Pola nama file yang akan dicopy
file_pattern = "temp_comments_"

# Cari dan salin file sesuai pola
for file_name in os.listdir(drive_folder):
    if file_name.startswith(file_pattern) and file_name.endswith(".csv"):
        source_path = os.path.join(drive_folder, file_name)
        destination_path = os.path.join(destination_folder, file_name)
        shutil.copy(source_path, destination_path)
        print(f"File '{file_name}' telah dicopy ke '{destination_folder}'.")

print("Proses copy selesai!")
```

```python
import os
import pandas as pd

# Folder tempat file komentar sementara disimpan
folder_path = "/content/tweets-data/"
output_filename = os.path.join(folder_path, "hasil_komentar.csv")

# Cari semua file yang sesuai pola temp_comments_*.csv
temp_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.startswith("temp_comments_") and f.endswith(".csv")]

# Gabungkan semua file
all_data = []
for temp_file in temp_files:
    try:
        temp_df = pd.read_csv(temp_file)
        all_data.append(temp_df)
        print(f"File {temp_file} berhasil dibaca.")
    except Exception as e:
        print(f"Gagal membaca file {temp_file}: {e}")

# Jika ada data, gabungkan menjadi satu DataFrame
if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)
    combined_df.to_csv(output_filename, index=False)
    print(f"Semua file berhasil digabungkan dan disimpan ke '{output_filename}'.")
else:
    print("Tidak ada data yang berhasil digabungkan.")
```

```python
import pandas as pd

# Path file hasil penggabungan
file_path = "/content/tweets-data/hasil_komentar.csv"  # Pastikan path sesuai dengan lokasi output sebelumnya

# Fungsi untuk membaca dan menampilkan data
def tampilkan_data(file_path, display_count=10):
    try:
        # Mengatur pandas untuk menampilkan semua kolom dan teks secara penuh
        pd.set_option('display.max_columns', None)  # Menampilkan semua kolom
        pd.set_option('display.max_colwidth', None)  # Menampilkan teks kolom penuh

        # Membaca file CSV hasil penggabungan
        df = pd.read_csv(file_path)  # Tidak perlu delimiter karena default-nya sudah koma
        print("Data hasil penggabungan:")

        # Menampilkan sejumlah baris sesuai parameter display_count
        display(df.head(display_count))

        # Menampilkan jumlah total baris
        print(f"Total baris dalam hasil: {len(df)}")
    except FileNotFoundError:
        print(f"File hasil penggabungan '{file_path}' tidak ditemukan.")
    except Exception as e:
        print(f"Terjadi kesalahan saat membaca file: {e}")

# Panggil fungsi dengan jumlah baris yang diinginkan
tampilkan_data(file_path, display_count=10)  # Ubah nilai display_count sesuai kebutuhan
```
